```python
# Additional Exercise 3: Transaction Analyzer

LAST_NAME = "Ayudtud"
SEED_NUM = "9"
FAVORITE_ARTIST = "IV of Spades"

# Generate five customer names
customers = [
    f"{LAST_NAME}_C{i}"
    for i in range(1, 6)
]

# Define the five required products
products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]

print("CUSTOMERS")
for customer in customers:
    print(customer)

print("\nPRODUCTS")
for product in products:
    print(product)
```

```python
# Step 2: Generate quantities

seed_value = int(SEED_NUM)

# Remove spaces so only artist-name characters are used
artist_characters = [char for char in FAVORITE_ARTIST if char != " "]

# Use character positions (A=1, B=2, ..., Z=26)
character_positions = [
    ord(char.upper()) - ord("A") + 1
    for char in artist_characters
    if char.isalpha()
]

# Generate quantities from the seed and character positions
quantities = [
    (seed_value + position) % 10 + 1
    for position in character_positions
]

print("ARTIST:", FAVORITE_ARTIST)
print("CHARACTER POSITIONS:", character_positions)
print("GENERATED QUANTITIES:", quantities)
```

```python
# Step 3: Assign fixed prices to products

product_prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 500,
    "CALCULATOR": 250
}

print("PRODUCT PRICES")
for product, price in product_prices.items():
    print(f"{product}: ₱{price}")
```

```python
# Step 4: Generate transaction records

transactions = []

for i in range(10):
    customer = customers[i % len(customers)]
    product = products[i % len(products)]
    quantity = quantities[i % len(quantities)]

    transaction = {
        "Customer": customer,
        "Product": product,
        "Quantity": quantity,
        "Unit Price": product_prices[product],
        "Transaction Value": quantity * product_prices[product]
    }

    transactions.append(transaction)

# Add a repeated customer-product purchase
transactions.append({
    "Customer": customers[0],
    "Product": products[0],
    "Quantity": quantities[0],
    "Unit Price": product_prices[products[0]],
    "Transaction Value": quantities[0] * product_prices[products[0]]
})

print("GENERATED TRANSACTION RECORDS")

for i, transaction in enumerate(transactions, start=1):
    print(f"\nTransaction {i}")
    for key, value in transaction.items():
        if key in ["Unit Price", "Transaction Value"]:
            print(f"{key}: ₱{value}")
        else:
            print(f"{key}: {value}")
```

```python
# Step 5: Identify repeated and distinct transactions

purchase_counts = {}

for transaction in transactions:
    key = (transaction["Customer"], transaction["Product"])
    purchase_counts[key] = purchase_counts.get(key, 0) + 1

print("REPEATED PURCHASES")

repeated_found = False

for (customer, product), count in purchase_counts.items():
    if count > 1:
        print(f"{customer} purchased {product} {count} times")
        repeated_found = True

if not repeated_found:
    print("No repeated purchases found.")

# Identify distinct customers and products
distinct_customers = sorted(set(
    transaction["Customer"] for transaction in transactions
))

distinct_products = sorted(set(
    transaction["Product"] for transaction in transactions
))

print("\nDISTINCT CUSTOMERS")
print(distinct_customers)

print("\nDISTINCT PRODUCTS")
print(distinct_products)
```

```python
# Step 6: Customer summaries

customer_summaries = {}

for customer in customers:
    customer_transactions = [
        transaction for transaction in transactions
        if transaction["Customer"] == customer
    ]

    customer_summaries[customer] = {
        "Number of Transactions": len(customer_transactions),
        "Total Quantity": sum(
            transaction["Quantity"]
            for transaction in customer_transactions
        ),
        "Total Spending": sum(
            transaction["Transaction Value"]
            for transaction in customer_transactions
        )
    }

print("CUSTOMER SUMMARIES")

for customer, summary in customer_summaries.items():
    print(f"\nCustomer: {customer}")
    print(f"Number of Transactions: {summary['Number of Transactions']}")
    print(f"Total Quantity: {summary['Total Quantity']}")
    print(f"Total Spending: ₱{summary['Total Spending']}")
```

```python
# Step 7: Product summaries

product_summaries = {}

for product in products:
    product_transactions = [
        transaction for transaction in transactions
        if transaction["Product"] == product
    ]

    product_summaries[product] = {
        "Number of Transactions": len(product_transactions),
        "Total Quantity Sold": sum(
            transaction["Quantity"]
            for transaction in product_transactions
        ),
        "Total Sales": sum(
            transaction["Transaction Value"]
            for transaction in product_transactions
        )
    }

print("PRODUCT SUMMARIES")

for product, summary in product_summaries.items():
    print(f"\nProduct: {product}")
    print(f"Number of Transactions: {summary['Number of Transactions']}")
    print(f"Total Quantity Sold: {summary['Total Quantity Sold']}")
    print(f"Total Sales: ₱{summary['Total Sales']}")
```

```python
# Step 8: Organize and sort transaction records

ordered_transactions = sorted(
    transactions,
    key=lambda transaction: (
        transaction["Customer"],
        transaction["Product"]
    )
)

print("ORDERED TRANSACTION RECORDS")

for transaction in ordered_transactions:
    print(
        f"Customer: {transaction['Customer']} | "
        f"Product: {transaction['Product']} | "
        f"Quantity: {transaction['Quantity']} | "
        f"Unit Price: ₱{transaction['Unit Price']} | "
        f"Transaction Value: ₱{transaction['Transaction Value']}"
    )
```

```python
# Step 9: Overall transaction summary

total_transactions = len(transactions)

total_quantity = sum(
    transaction["Quantity"]
    for transaction in transactions
)

overall_transaction_value = sum(
    transaction["Transaction Value"]
    for transaction in transactions
)

total_customers = len(distinct_customers)
total_products = len(distinct_products)

print("OVERALL TRANSACTION SUMMARY")
print(f"Total Transactions: {total_transactions}")
print(f"Distinct Customers: {total_customers}")
print(f"Distinct Products: {total_products}")
print(f"Total Quantity Purchased: {total_quantity}")
print(f"Overall Transaction Value: ₱{overall_transaction_value}")
```

```python
# Step 10: Final Consolidated Transaction Report

print("=" * 60)
print("       FINAL CONSOLIDATED TRANSACTION REPORT")
print("=" * 60)

# 1. Ordered Transaction Records
print("\n1. ORDERED TRANSACTION RECORDS")
for transaction in ordered_transactions:
    print(
        f"{transaction['Customer']} | "
        f"{transaction['Product']} | "
        f"Quantity: {transaction['Quantity']} | "
        f"Unit Price: ₱{transaction['Unit Price']} | "
        f"Value: ₱{transaction['Transaction Value']}"
    )

# 2. Repeated Purchases
print("\n2. REPEATED PURCHASES")
for (customer, product), count in purchase_counts.items():
    if count > 1:
        print(f"{customer} - {product}: {count} transactions")

# 3. Customer Summaries
print("\n3. CUSTOMER SUMMARIES")
for customer, summary in customer_summaries.items():
    print(
        f"{customer}: "
        f"{summary['Number of Transactions']} transactions, "
        f"Quantity: {summary['Total Quantity']}, "
        f"Spending: ₱{summary['Total Spending']}"
    )

# 4. Product Summaries
print("\n4. PRODUCT SUMMARIES")
for product, summary in product_summaries.items():
    print(
        f"{product}: "
        f"{summary['Number of Transactions']} transactions, "
        f"Quantity Sold: {summary['Total Quantity Sold']}, "
        f"Sales: ₱{summary['Total Sales']}"
    )

# 5. Distinct Customers and Products
print("\n5. DISTINCT CUSTOMERS")
print(", ".join(distinct_customers))

print("\nDISTINCT PRODUCTS")
print(", ".join(distinct_products))

# 6. Overall Summary
print("\n6. OVERALL TRANSACTION SUMMARY")
print(f"Total Transactions: {total_transactions}")
print(f"Total Customers: {total_customers}")
print(f"Total Products: {total_products}")
print(f"Total Quantity Purchased: {total_quantity}")
print(f"Overall Transaction Value: ₱{overall_transaction_value}")

print("\n" + "=" * 60)
print("             END OF TRANSACTION REPORT")
print("=" * 60)
```
