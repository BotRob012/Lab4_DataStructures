```python
# Additional Exercise 3: Transaction Analyzer

LAST_NAME = "Ayudtud"
SEED_NUM = "9"
FAVORITE_ARTIST = "IV of Spades"

# Generate five customer names
customers = [
    f"{LAST_NAME}_C{i}"
    for i in range(1, 6)
]

# Define the five required products
products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]

print("CUSTOMERS")
for customer in customers:
    print(customer)

print("\nPRODUCTS")
for product in products:
    print(product)
```

```python
# Step 2: Generate quantities

seed_value = int(SEED_NUM)

# Remove spaces so only artist-name characters are used
artist_characters = [char for char in FAVORITE_ARTIST if char != " "]

# Use character positions (A=1, B=2, ..., Z=26)
character_positions = [
    ord(char.upper()) - ord("A") + 1
    for char in artist_characters
    if char.isalpha()
]

# Generate quantities from the seed and character positions
quantities = [
    (seed_value + position) % 10 + 1
    for position in character_positions
]

print("ARTIST:", FAVORITE_ARTIST)
print("CHARACTER POSITIONS:", character_positions)
print("GENERATED QUANTITIES:", quantities)
```
