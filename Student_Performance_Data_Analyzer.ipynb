```python
# Exercise 2: Student Performance Analyzer

# Required student-specific inputs
LAST_NAME = "Ayudtud"
SEED_NUM = "9"
FAVORITE_ARTIST = "IV of Spades"

# Generate five student names
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

# Define the five required subjects
subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

print("STUDENTS")
for student in students:
    print(student)

print("\nSUBJECTS")
for subject in subjects:
    print(subject)
```

```python
# Generate performance scores using SEED_NUM
seed_value = int(SEED_NUM)

performance_scores = [
    seed_value + position
    for position in range(len(FAVORITE_ARTIST))
]

print("Favorite Artist:", FAVORITE_ARTIST)
print("Generated Performance Scores:", performance_scores)
```

```python
# Step 3: Generate student performance records

performance_records = []

for student_index, student in enumerate(students):
    for subject_index, subject in enumerate(subjects):
        score = int(SEED_NUM) + student_index + subject_index

        performance_records.append({
            "Student": student,
            "Subject": subject,
            "Score": score
        })

# Add one repeated student-subject record
performance_records.append(performance_records[0].copy())

# Add one incomplete record with a missing score
performance_records.append({
    "Student": students[0],
    "Subject": subjects[1],
    "Score": None
})

print("Generated Performance Records:")
for record in performance_records:
    print(record)
```

```python
# Step 4: Organize student and subject performance data

organized_records = sorted(
    performance_records,
    key=lambda record: (record["Student"], record["Subject"])
)

print("Organized Student/Subject Data:")
for record in organized_records:
    print(record)
```

```python
# Step 5: Identify repeated and incomplete records

seen_records = set()
repeated_records = []
incomplete_records = []

for record in performance_records:
    record_key = (
        record["Student"],
        record["Subject"],
        record["Score"]
    )

    if record_key in seen_records:
        repeated_records.append(record)
    else:
        seen_records.add(record_key)

    if any(value is None for value in record.values()):
        incomplete_records.append(record)

print("Repeated Records:")
for record in repeated_records:
    print(record)

print("\nIncomplete Records:")
for record in incomplete_records:
    print(record)
```

```python
# Step 6: Determine performance categories

def get_performance_category(score):
    if score is None:
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Satisfactory"
    else:
        return "Needs Improvement"


for record in performance_records:
    record["Category"] = get_performance_category(record["Score"])

print("Performance Categories:")
for record in performance_records:
    print(record)
```

```python
# Step 7: Calculate student and subject summaries

student_summaries = {}
subject_summaries = {}

# Summarize each student's scores
for student in students:
    scores = [
        record["Score"]
        for record in performance_records
        if record["Student"] == student
        and record["Score"] is not None
    ]

    student_summaries[student] = {
        "Average": round(sum(scores) / len(scores), 2),
        "Highest": max(scores),
        "Lowest": min(scores)
    }

# Summarize each subject's scores
for subject in subjects:
    scores = [
        record["Score"]
        for record in performance_records
        if record["Subject"] == subject
        and record["Score"] is not None
    ]

    subject_summaries[subject] = {
        "Average": round(sum(scores) / len(scores), 2),
        "Highest": max(scores),
        "Lowest": min(scores)
    }

print("STUDENT SUMMARIES")
for student, summary in student_summaries.items():
    print(student, summary)

print("\nSUBJECT SUMMARIES")
for subject, summary in subject_summaries.items():
    print(subject, summary)
```

```python
# Step 9: Final Student Performance Report

# Re-create ordered summaries
ordered_student_summaries = dict(sorted(student_summaries.items()))
ordered_subject_summaries = dict(sorted(subject_summaries.items()))

print("=" * 60)
print("       STUDENT PERFORMANCE ANALYZER REPORT")
print("=" * 60)

print("\n1. GENERATED PERFORMANCE RECORDS")
print("Total records:", len(performance_records))

print("\n2. REPEATED AND INCOMPLETE RECORDS")
print("Repeated records:", len(repeated_records))
print("Incomplete records:", len(incomplete_records))

print("\n3. PERFORMANCE CATEGORIES")
category_counts = {}

for record in performance_records:
    category = record["Category"]
    category_counts[category] = category_counts.get(category, 0) + 1

for category, count in sorted(category_counts.items()):
    print(f"{category}: {count}")

print("\n4. STUDENT PERFORMANCE SUMMARIES")
for student, summary in ordered_student_summaries.items():
    print(
        f"{student}: Average = {summary['Average']}, "
        f"Highest = {summary['Highest']}, "
        f"Lowest = {summary['Lowest']}"
    )

print("\n5. SUBJECT PERFORMANCE SUMMARIES")
for subject, summary in ordered_subject_summaries.items():
    print(
        f"{subject}: Average = {summary['Average']}, "
        f"Highest = {summary['Highest']}, "
        f"Lowest = {summary['Lowest']}"
    )

print("\n6. FINAL STATUS")
print("Performance analysis completed.")
print("=" * 60)
```
